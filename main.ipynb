```python
import pandas as pd
import os

def load_data():
    py_path = os.getcwd()
    CMS_path = os.path.join(py_path, 'CMS.CSV')

    CMS_Data = pd.read_csv(CMS_path, encoding='utf-8')
    return CMS_Data

CMS_data = load_data()
```

```python
def initial_data(data):
    #Check columns names
    print('Columns Names:')
    for col in data.columns:
        print({col})

    print(f"Data Shape: {data.shape}")

initial_data(CMS_data)
```
